```python
### PREPARACIÓN DEL ENTORNO
import plotly.express as px
import pandas as pd

### CARGA DE DATOS
pelis = pd.read_csv("movies.csv")
```

```python
# -------- Gráfico de barras -------- Cantidad de películas por género
def pelis_por_genero(pelis):
    generos = (
        pelis["genres"].fillna("").str.split(",").explode().str.strip()
    )  # Separar los géneros que están unidos por comas

    generos = generos[generos != ""]  # Eliminar los géneros vacíos

    cantidad_por_genero = generos.value_counts().reset_index()

    figura_generos = px.bar(  # Crear la gráfica
        cantidad_por_genero,
        y="genres",
        x="count",
        title="🎭 Análisis por género",
        labels={"count": "Cantidad de películas", "genres": ""},
    )

    return figura_generos  # Devolver la gráfica
```
